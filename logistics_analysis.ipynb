# Hypothetical logistics dataset

```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Shipment_ID":[f"S{i:03d}" for i in range(1,31)],
    "Distance_km":np.random.randint(100,1000,30),
    "Shipment_Volume_tons":np.random.randint(5,30,30),
    "Delivery_Time_hours":np.random.randint(4,30,30),
    "Transportation_Cost":np.random.randint(5000,40000,30)
})

```

```python
print("Average Delivery Time:", df["Delivery_Time_hours"].mean())
print("Average Transportation Cost:", df["Transportation_Cost"].mean())
```

# Correlation

```python
print(df.corr(numeric_only=True))
```

# Visualization

# Distance vs Cost
The scatter plot shows the relationship between shipment distance and transportation cost. The data indicates that transportation costs vary across shipments, with higher distances generally associated with higher costs. This visualization helps identify distance as an important factor affecting logistics expenses.

```python
import matplotlib.pyplot as plt

plt.scatter(df["Distance_km"], df["Transportation_Cost"])
plt.title("Distance vs Transportation Cost")
plt.xlabel("Distance (km)")
plt.ylabel("Transportation Cost")
plt.show()
```

# Delivery Time by Shipment
The bar chart compares delivery times across all shipments. Delivery times range from approximately 5 to 29 hours. Some shipments require significantly more time than others, which may be due to route length, transportation conditions, or operational factors.

```python
plt.bar(df["Shipment_ID"], df["Delivery_Time_hours"])
plt.title("Delivery Time by Shipment")
plt.xlabel("Shipment ID")
plt.ylabel("Delivery Time (hours)")
plt.show()
```

# Distribution of Delivery Time
Most deliveries are concentrated in the lower delivery-time range, while a few shipments take considerably longer.

```python
plt.hist(df["Delivery_Time_hours"], bins=5)
plt.title("Distribution of Delivery Time")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()
```
